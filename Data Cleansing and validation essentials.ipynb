```python
import pandas as pd
from sqlalchemy import create_engine, text
from urllib.parse import quote_plus

# ====================================================
# SQL Server Connection
# ====================================================

conn_str = quote_plus(
    "DRIVER={ODBC Driver 17 for SQL Server};"
    "SERVER=SLIVAOLA\\SQL_SERVER_DEV;"
    "DATABASE=Disorders;"
    "UID=sa;"
    "PWD=Persik391@;"
)

engine = create_engine(
    f"mssql+pyodbc:///?odbc_connect={conn_str}"
)

# ====================================================
# Test Connection
# ====================================================

with engine.connect() as conn:
    result = conn.execute(text("SELECT DB_NAME()"))
    print("Connected to:", result.fetchone()[0])

# ====================================================
# Read CSV File
# ====================================================

df = pd.read_csv(
    r"C:\Users\natal\Documents\loan_data.csv"
)

# ====================================================
# Optional Data Validation
# ====================================================

print("Rows:", len(df))
print(df.dtypes)

# Remove duplicate rows
df = df.drop_duplicates()

# ====================================================
# Load to SQL Server
# ====================================================

df.to_sql(
    name="stg_loan_data",
    con=engine,
    schema="dbo",
    if_exists="replace",     # replace | append | fail
    index=False,
    chunksize=5000,
    method="multi"
)

print("Load completed successfully")

# ====================================================
# Verify Row Count
# ====================================================

with engine.connect() as conn:
    result = conn.execute(
        text("SELECT COUNT(*) FROM dbo.stg_loan_data")
    )
    print("Rows loaded:", result.scalar())
```
