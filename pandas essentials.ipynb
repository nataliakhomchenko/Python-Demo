```python
import pandas as pd

# Load data
df = pd.read_csv("c:/datasets/Customers.csv")

# Filter rows based on conditions
filtered_df = df[(df["Marital_Status"] == "M") & (df["Total_Children"] > 1)]

# Select specific columns and handle missing values
df_subset = df[["Customer_Name", "Birth_Date"]].dropna()

# Conditional column creation (like an IF statement)
import numpy as np
df["Category"] = np.where(df["Total_Children"] > 1, "High", "Standard")
```

```python
df.head()
```

```python
# Groupby with multiple aggregations
summary = df.groupby("Marital_Status").agg(
    Total_Children=("Total_Children", "sum"),
    Number_of_Children_at_Home=("Number_of_Children_at_Home", "mean"),
    Number_of_Educations=("Education", "count")
).reset_index()
```

```python
summary.head()
```

```python
# Joins (Merging) dataframes on a key
merged_df = pd.merge(summary, df, on="Marital_Status", how="left")
```
