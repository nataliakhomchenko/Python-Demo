```python
import duckdb

query = """
SELECT
    Model,
    SUM(Sales_Volume) AS TotalSales
FROM read_csv_auto('C:/datasets/BMW sales.csv')
GROUP BY Model
ORDER BY TotalSales DESC
"""

result = duckdb.sql(query).df()
print(result)
```

```python
import duckdb

query = """
SELECT
    Model,
    Sales,
    ROW_NUMBER() OVER (
        PARTITION BY Model
        ORDER BY Sales DESC
    ) AS RowNum
FROM read_csv_auto('C:/datasets/BMW sales.csv')
"""

df = duckdb.sql(query).df()

print(df.head())
```
