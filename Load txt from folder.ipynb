```python
#load the first 100 records from each file in the ClinicalTrials.gov export folder
from pathlib import Path
import pandas as pd

folder = Path(r"C:\datasets\clinical_trials\export_ctgov")

all_dfs = []

for file in folder.glob("*.txt"):
    try:
        # Try pipe delimiter first
        try:
            df = pd.read_csv(
                file,
                sep="|",
                nrows=100,
                low_memory=False
            )
            
            # If only one column was read, try tab delimiter
            if len(df.columns) == 1:
                raise ValueError("Wrong delimiter")

        except:
            df = pd.read_csv(
                file,
                sep="\t",
                nrows=100,
                low_memory=False
            )

        df["source_file"] = file.stem

        all_dfs.append(df)

        print(f"Loaded {len(df):3d} rows from {file.name}")

    except Exception as e:
        print(f"Failed: {file.name}")
        print(f"Error: {e}")

print("\nFiles loaded:", len(all_dfs))

# Optional combined dataset
combined_df = pd.concat(
    all_dfs,
    ignore_index=True,
    sort=False
)

print("Total rows:", len(combined_df))
print("Total columns:", len(combined_df.columns))

combined_df.head()
```

```python
"""
Reads every .txt file in C:\datasets\clinical_trials\export_ctgov
Creates the folder C:\datasets\clinical_trials\export_ctgov\parquet if it doesn't exist
Converts each file to a separate Parquet file
Preserves the original filename
Just add nrows=100 when reading the TXT file.
"""

from pathlib import Path
import pandas as pd

source_dir = Path(r"C:\datasets\clinical_trials\export_ctgov")
target_dir = source_dir / "parquet"

target_dir.mkdir(exist_ok=True)

for txt_file in source_dir.glob("*.txt"):

    try:
        df = pd.read_csv(
            txt_file,
            sep="|",           # change to '\t' if tab-delimited
            nrows=100,         # only first 100 records
            low_memory=False
        )

        parquet_file = target_dir / f"{txt_file.stem}.parquet"

        df.to_parquet(
            parquet_file,
            engine="pyarrow",
            compression="snappy",
            index=False
        )

        print(
            f"Created {parquet_file.name} "
            f"with {len(df)} records"
        )

    except Exception as e:
        print(f"Failed: {txt_file.name}")
        print(e)

print("Done.")
```
